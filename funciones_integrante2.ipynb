## 1. `sum()`
**¿Qué hace?** Suma todos los elementos numéricos de un iterable (lista, tupla, etc.).

**Sintaxis:** `sum(iterable, inicio=0)`

```python
# Ejemplo: total de tres pensiones reconocidas (en pesos)
pensiones = [1300000, 1500000, 1800000]
print("Total pensiones:", sum(pensiones))
```

## 2. `min()`
**¿Qué hace?** Devuelve el valor más pequeño de un iterable o de varios argumentos.

**Sintaxis:** `min(iterable)` o `min(a, b, c, ...)`

```python
# Ejemplo: el predio de menor extensión (en hectáreas)
hectareas = [12.5, 3.2, 45.0, 8.7]
print("Predio más pequeño:", min(hectareas), "ha")
```

## 3. `max()`
**¿Qué hace?** Devuelve el valor más grande de un iterable o de varios argumentos.

**Sintaxis:** `max(iterable)` o `max(a, b, c, ...)`

```python
# Ejemplo: el predio de mayor extensión
hectareas = [12.5, 3.2, 45.0, 8.7]
print("Predio más grande:", max(hectareas), "ha")
```

## 4. `abs()`
**¿Qué hace?** Devuelve el valor absoluto de un número (lo convierte en positivo).

**Sintaxis:** `abs(numero)`

```python
# Ejemplo: diferencia entre el valor solicitado y el reconocido en una indemnización
solicitado = 5000000
reconocido = 6200000
diferencia = solicitado - reconocido
print("Diferencia:", diferencia)
print("Diferencia en valor absoluto:", abs(diferencia))
```

## 5. `round()`
**¿Qué hace?** Redondea un número a la cantidad de decimales indicada (0 por defecto).

**Sintaxis:** `round(numero, decimales)`

```python
# Ejemplo: porcentaje de cobertura de una población campesina
porcentaje = 73.46829
print("Sin decimales:", round(porcentaje))
print("Con 2 decimales:", round(porcentaje, 2))
```

## 6. `pow()`
**¿Qué hace?** Eleva un número a una potencia. Equivale al operador `**`.

**Sintaxis:** `pow(base, exponente)`

```python
# Ejemplo: crecimiento de una deuda con un factor de 1,02 durante 5 periodos
factor = 1.02
periodos = 5
print("Factor acumulado:", round(pow(factor, periodos), 4))
print("2 elevado a la 10:", pow(2, 10))
```

## 7. `divmod()`
**¿Qué hace?** Devuelve una tupla con el cociente y el residuo de una división entera.

**Sintaxis:** `divmod(a, b)`

```python
# Ejemplo: convertir 400 días de trámite en años y días restantes
dias = 400
anios, dias_restantes = divmod(dias, 365)
print(f"{dias} días = {anios} año(s) y {dias_restantes} día(s)")
```

## 8. `sorted()`
**¿Qué hace?** Devuelve una **nueva lista ordenada** a partir de un iterable, sin modificar el original.

**Sintaxis:** `sorted(iterable, reverse=False)`

```python
# Ejemplo: ordenar leyes por año de expedición
leyes = [1994, 1993, 2011, 1991]
print("Ascendente:", sorted(leyes))
print("Descendente:", sorted(leyes, reverse=True))
print("Original sin cambios:", leyes)
```

## 9. `reversed()`
**¿Qué hace?** Devuelve un iterador que recorre los elementos en orden inverso. Para verlo como lista se usa `list()`.

**Sintaxis:** `reversed(secuencia)`

```python
# Ejemplo: instancias de un proceso, de la última a la primera
instancias = ["Primera instancia", "Segunda instancia", "Casación"]
print(list(reversed(instancias)))
```

## 10. `map()`
**¿Qué hace?** Aplica una función a **cada elemento** de un iterable y devuelve los resultados.

**Sintaxis:** `map(funcion, iterable)`

```python
# Ejemplo: convertir salarios en SMMLV (suponiendo un SMMLV de 1.300.000)
salarios = [2600000, 3900000, 1300000]
en_smmlv = list(map(lambda s: s / 1300000, salarios))
print("Salarios en SMMLV:", en_smmlv)
```

## 11. `filter()`
**¿Qué hace?** Filtra los elementos de un iterable que cumplen una condición (la función debe devolver `True` o `False`).

**Sintaxis:** `filter(funcion, iterable)`

```python
# Ejemplo: quedarse con los predios de más de 10 hectáreas
hectareas = [12.5, 3.2, 45.0, 8.7]
grandes = list(filter(lambda h: h > 10, hectareas))
print("Predios de más de 10 ha:", grandes)
```

## 12. `any()`
**¿Qué hace?** Devuelve `True` si **al menos un** elemento del iterable es verdadero.

**Sintaxis:** `any(iterable)`

```python
# Ejemplo: ¿al menos uno de los requisitos de la solicitud está cumplido?
requisitos = [False, True, False]
print("¿Algún requisito cumplido?", any(requisitos))
```

## 13. `all()`
**¿Qué hace?** Devuelve `True` solo si **todos** los elementos del iterable son verdaderos.

**Sintaxis:** `all(iterable)`

```python
# Ejemplo: ¿se cumplieron todos los requisitos para reconocer la pensión?
requisitos = [True, True, True]
print("¿Todos cumplidos?", all(requisitos))

requisitos_incompletos = [True, False, True]
print("¿Todos cumplidos (caso incompleto)?", all(requisitos_incompletos))
```

## 14. `isinstance()`
**¿Qué hace?** Comprueba si un objeto es de un tipo determinado. Devuelve `True` o `False`.

**Sintaxis:** `isinstance(objeto, tipo)`

```python
# Ejemplo: validar el tipo de dato de un campo antes de procesarlo
hectareas = 12.5
cedula = "1020304050"
print("¿hectareas es float?", isinstance(hectareas, float))
print("¿cedula es int?", isinstance(cedula, int))
print("¿cedula es str?", isinstance(cedula, str))
```

## 15. `help()`

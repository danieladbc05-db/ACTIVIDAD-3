## 1. `print()`
**¿Qué hace?** Muestra un mensaje o valor en pantalla.

**Sintaxis:** `print(valor1, valor2, ...)`

```python
# Ejemplo: mostrar datos de un expediente
print("Expediente 2024-0158: Restitución de tierras")
print("Predios:", 3, "| Hectáreas:", 45.5)
```

## 2. `input()`
**¿Qué hace?** Pide al usuario un dato por teclado y lo devuelve como texto (`str`).

**Sintaxis:** `input("mensaje")`

```python
# Ejemplo: registrar el nombre del demandante (aparece una cajita para escribir)
demandante = input("Nombre del demandante: ")
print("Registrado:", demandante)
```

## 3. `len()`
**¿Qué hace?** Devuelve la cantidad de elementos de un objeto (lista, texto, diccionario...).

**Sintaxis:** `len(objeto)`

```python
# Ejemplo: contar normas en una lista y letras en un texto
normas = ["Ley 100 de 1993", "Ley 160 de 1994", "Ley 1448 de 2011"]
print("Cantidad de normas:", len(normas))
print("Letras en 'Constitución':", len("Constitución"))
```

## 4. `type()`
**¿Qué hace?** Devuelve el tipo de dato de un valor.

**Sintaxis:** `type(objeto)`

```python
# Ejemplo: identificar el tipo de cada dato de un predio
hectareas = 12.5
matricula = "050-123456"
num_propietarios = 3
print(type(hectareas))
print(type(matricula))
print(type(num_propietarios))
```

## 5. `range()`
**¿Qué hace?** Genera una secuencia de números enteros. Se usa mucho en ciclos `for`.

**Sintaxis:** `range(inicio, fin, paso)` (el `fin` no se incluye)

```python
# Ejemplo: años de radicación de un proceso
for anio in range(2020, 2024):
    print("Radicación año", anio)

print("Pares del 0 al 8:", list(range(0, 10, 2)))
```

## 6. `str()`
**¿Qué hace?** Convierte un valor a texto.

**Sintaxis:** `str(valor)`

```python
# Ejemplo: armar el nombre de una ley a partir de números
numero = 160
anio = 1994
print("Ley " + str(numero) + " de " + str(anio))
```

## 7. `int()`
**¿Qué hace?** Convierte un valor a número entero (si es decimal, descarta los decimales).

**Sintaxis:** `int(valor)`

```python
# Ejemplo: convertir texto a número para poder operar
dato = "45"
print(int(dato) + 5)
print(int(12.9))
```

## 8. `float()`
**¿Qué hace?** Convierte un valor a número decimal.

**Sintaxis:** `float(valor)`

```python
# Ejemplo: extensión de un predio que llegó como texto
extension = "12.5"
print(float(extension) * 2)
print(float(3))
```

## 9. `bool()`
**¿Qué hace?** Convierte un valor a `True` o `False`. El 0, el texto vacío y las listas vacías dan `False`.

**Sintaxis:** `bool(valor)`

```python
# Ejemplo: validar si un campo del formulario tiene contenido
print(bool(0))
print(bool(5))
print(bool(""))
print(bool("Ley 100"))
```

## 10. `list()`
**¿Qué hace?** Crea una lista a partir de un iterable. Las listas se pueden modificar.

**Sintaxis:** `list(iterable)`

```python
# Ejemplo: crear una lista de normas y agregar una más
normas = list(("Ley 100 de 1993", "Ley 160 de 1994"))
normas.append("Ley 1448 de 2011")
print(normas)
```

## 11. `tuple()`
**¿Qué hace?** Crea una tupla, parecida a una lista pero que **no se puede modificar**.

**Sintaxis:** `tuple(iterable)`

```python
# Ejemplo: coordenadas de un predio (no deben cambiar)
coordenadas = tuple([4.711, -74.072])
print(coordenadas)
print("Latitud:", coordenadas[0])
```

## 12. `dict()`
**¿Qué hace?** Crea un diccionario: un conjunto de pares `clave: valor`.

**Sintaxis:** `dict(clave=valor, ...)`

```python
# Ejemplo: ficha de un predio
predio = dict(matricula="050-123456", area_ha=12.5, municipio="Sopó")
print(predio)
print("Municipio:", predio["municipio"])
```

## 13. `set()`
**¿Qué hace?** Crea un conjunto: una colección **sin elementos repetidos**.

**Sintaxis:** `set(iterable)`

```python
# Ejemplo: eliminar radicados repetidos
radicados = ["R-001", "R-002", "R-001", "R-003", "R-002"]
unicos = set(radicados)
print(unicos)
print("Radicados únicos:", len(unicos))
```

## 14. `enumerate()`
**¿Qué hace?** Recorre un iterable devolviendo el número de posición junto con cada elemento.

**Sintaxis:** `enumerate(iterable, start=0)`

```python
# Ejemplo: numerar las normas de una lista
normas = ["Ley 100 de 1993", "Ley 160 de 1994", "Ley 1448 de 2011"]
for numero, norma in enumerate(normas, start=1):
    print(numero, "-", norma)
```

## 15. `zip()`
**¿Qué hace?** Une varios iterables elemento por elemento, formando pares (o tuplas).

**Sintaxis:** `zip(iterable1, iterable2, ...)`

```python
# Ejemplo: emparejar cada predio con su extensión
predios = ["La Esperanza", "El Roble", "San José"]
hectareas = [12.5, 3.2, 45.0]
for predio, ha in zip(predios, hectareas):
    print(predio, "->", ha, "ha")
```
